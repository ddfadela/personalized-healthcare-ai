```python
import os

data_folder = "../data"
mimic_folder = None

if os.path.exists(data_folder):
    for name in os.listdir(data_folder):
        if "mimic" in name.lower():
            mimic_folder = os.path.join(data_folder, name)
            break

if mimic_folder and os.path.exists(mimic_folder):
    for dirname, _, filenames in os.walk(mimic_folder):
        print("Directory:", dirname)
        for filename in filenames:
            print(os.path.join(dirname, filename))

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pathlib
import warnings
admissions = pd.read_csv('../data/mimic-iii-clinical-database-demo-1.4/ADMISSIONS.csv')
patients = pd.read_csv('../data/mimic-iii-clinical-database-demo-1.4/PATIENTS.csv')
prescriptions = pd.read_csv('../data/mimic-iii-clinical-database-demo-1.4/PRESCRIPTIONS.csv')
labevents = pd.read_csv('../data/mimic-iii-clinical-database-demo-1.4/LABEVENTS.csv')
chartevents = pd.read_csv(
    '../data/mimic-iii-clinical-database-demo-1.4/CHARTEVENTS.csv',
    low_memory=False
)

diagnoses = pd.read_csv('../data/mimic-iii-clinical-database-demo-1.4/D_ICD_DIAGNOSES.csv')
```

```python
def info_summary(df, name):
    print(f"✅ {name}")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("-" * 50)

info_summary(admissions, "ADMISSIONS")
info_summary(patients, "PATIENTS")
info_summary(prescriptions, "PRESCRIPTIONS")
info_summary(labevents, "LABEVENTS")
info_summary(chartevents, "CHARTEVENTS")
info_summary(diagnoses, "DIAGNOSES_ICD")
```

```python
# Merge demographics and admission data
patient_adm = pd.merge(patients, admissions, on='subject_id')

# Explore most common drugs
top_drugs = prescriptions['drug'].value_counts().head(10)

# Check available labs
top_labs = labevents['itemid'].value_counts().head(10)

print('top drugs',top_drugs)
print('top labs',top_labs)
```
